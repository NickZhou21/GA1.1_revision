```python
def sum_list(values):
    total = 0
    for value in values:
        total = add(value, total)
    return total
```

```python
sum_list([1, 2, 3, 4, 5])
```

```python
sum_list([1, 2, 3, 4, 2, 3, 4, 5, 6, 7, 8, 9, 10])
```

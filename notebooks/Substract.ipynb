```python
def substract(a,b):
    return a - b
```

```python
substract(1, 2)
```

```python
substract(3, 2)
```

```python
substract(1, 10)
```

```python
def distance_from_zero(x):
    if x < 0:
        return substract(0, x)
    else:
        return substract(x, 0)
```

```python
distance_from_zero(9)
```

```python
distance_from_zero(3)
```

```python
distance_from_zero(4)
```
